```python

# ============================================================
# INTERNCIRCLE - DATA VISUALIZATION PROJECT
# Titanic Dataset | Matplotlib, Seaborn & Pandas
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. SET VISUALIZATION STYLE
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams["figure.dpi"] = 120
plt.rcParams["axes.titleweight"] = "bold"
plt.rcParams["axes.titlesize"] = 15

print("=" * 60)
print("TITANIC DATA VISUALIZATION PROJECT")
print("=" * 60)

# 3. LOAD DATASET
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)

print("\nDataset loaded successfully!")
print("Original dataset shape:", df.shape)

# 4. EXPLORE DATASET
print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

# 5. DATA CLEANING
# Keep all passenger records; do not remove rows
# just because their recorded values are identical.
df_viz = df.copy()

print("\nMissing Values Before Cleaning:")
print(df_viz.isnull().sum())

print("\nExact Duplicate Rows (not removed):",
      df_viz.duplicated().sum())

# Fill missing age with median
df_viz["age"] = df_viz["age"].fillna(
    df_viz["age"].median()
)

# Fill missing embarkation port with mode
df_viz["embarked"] = df_viz["embarked"].fillna(
    df_viz["embarked"].mode()[0]
)

print("\nDataset Shape After Cleaning:", df_viz.shape)
print("\nMissing Values After Cleaning:")
print(df_viz.isnull().sum())

# 6. HISTOGRAM - AGE DISTRIBUTION
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df_viz,
    x="age",
    bins=20,
    color="royalblue",
    edgecolor="white"
)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()

# 7. BAR CHART - SURVIVAL RATE BY GENDER
survival_gender = (
    df_viz.groupby("sex")["survived"].mean() * 100
)

plt.figure(figsize=(8, 5))

plt.bar(
    survival_gender.index,
    survival_gender.values,
    color=["#e78ac3", "#66c2a5"],
    edgecolor="black"
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)

for i, value in enumerate(survival_gender.values):
    plt.text(i, value + 2, f"{value:.1f}%",
             ha="center")

plt.tight_layout()
plt.show()

# 8. BAR CHART - SURVIVAL RATE BY CLASS
survival_class = (
    df_viz.groupby("pclass")["survived"].mean() * 100
)

plt.figure(figsize=(8, 5))

plt.bar(
    survival_class.index.astype(str),
    survival_class.values,
    color=["#4daf4a", "#377eb8", "#ff7f00"],
    edgecolor="black"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)

for i, value in enumerate(survival_class.values):
    plt.text(i, value + 2, f"{value:.1f}%",
             ha="center")

plt.tight_layout()
plt.show()

# 9. SCATTER PLOT - AGE VS FARE
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df_viz,
    x="age",
    y="fare",
    hue="survived",
    palette={0: "#e76f51", 1: "#2a9d8f"},
    alpha=0.7
)

plt.title("Passenger Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.tight_layout()
plt.show()

# 10. BOX PLOT - FARE BY CLASS
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df_viz,
    x="pclass",
    y="fare",
    hue="pclass",
    palette="Set2",
    legend=False
)

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()

# 11. CORRELATION MATRIX
numeric_df = df_viz.select_dtypes(include="number")
correlation_matrix = numeric_df.corr()

print("\nCorrelation Matrix:")
display(correlation_matrix.round(2))

# 12. CORRELATION HEATMAP
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    center=0
)

plt.title("Correlation Heatmap of Titanic Dataset")
plt.tight_layout()
plt.show()

# 13. SURVIVAL COUNT CHART
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_viz,
    x="survived",
    hue="survived",
    palette={0: "#e76f51", 1: "#2a9d8f"},
    legend=False
)

plt.title("Survival Count of Titanic Passengers")
plt.xlabel("Survival Status (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.xticks([0, 1], ["Did Not Survive", "Survived"])
plt.tight_layout()
plt.show()

# 14. STATISTICAL ANALYSIS
print("\n" + "=" * 60)
print("SURVIVAL ANALYSIS")
print("=" * 60)

print("\nSurvival by Gender:")
gender_analysis = df_viz.groupby("sex")["survived"].agg(
    Passengers="count",
    Survivors="sum",
    Survival_Rate="mean"
)
gender_analysis["Survival_Rate"] *= 100
display(gender_analysis.round(2))

print("\nSurvival by Passenger Class:")
class_analysis = df_viz.groupby("pclass")["survived"].agg(
    Passengers="count",
    Survivors="sum",
    Survival_Rate="mean"
)
class_analysis["Survival_Rate"] *= 100
display(class_analysis.round(2))

print("\nAverage Age and Fare by Survival Status:")
age_fare_analysis = df_viz.groupby("survived").agg(
    Average_Age=("age", "mean"),
    Average_Fare=("fare", "mean"),
    Passenger_Count=("survived", "count")
)
display(age_fare_analysis.round(2))

# 15. AUTOMATIC KEY FINDINGS
print("\n" + "=" * 60)
print("KEY FINDINGS")
print("=" * 60)

print("Total passengers:", len(df_viz))

print(
    f"Overall survival rate: "
    f"{df_viz['survived'].mean() * 100:.2f}%"
)

print(
    f"Average passenger age: "
    f"{df_viz['age'].mean():.2f} years"
)

print(
    f"Average ticket fare: "
    f"{df_viz['fare'].mean():.2f}"
)

print(
    f"Gender with higher survival rate: "
    f"{survival_gender.idxmax().capitalize()} "
    f"({survival_gender.max():.2f}%)"
)

print(
    f"Passenger class with highest survival rate: "
    f"{survival_class.idxmax()} "
    f"({survival_class.max():.2f}%)"
)

print("\nCorrelation with Survival:")
print(correlation_matrix["survived"].sort_values(
    ascending=False
).round(2))

# 16. PROJECT CONCLUSION
print("\n" + "=" * 60)
print("PROJECT CONCLUSION")
print("=" * 60)

print("""
The Titanic dataset was explored and visualized using
Pandas, Matplotlib and Seaborn.

Histograms, bar charts, scatter plots, box plots and
a correlation heatmap were created to understand
passenger distributions and relationships.

The analysis demonstrates how data visualization
can help communicate patterns and statistical insights
from a real-world dataset.

PROJECT COMPLETED SUCCESSFULLY!
""")

```
