```python
import gradio as gr
import numpy as np
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier


iris = load_iris()
X, y = iris.data, iris.target
model = RandomForestClassifier(random_state=42)
model.fit(X, y)
target_names = ["Setosa", "Versicolor", "Virginica"]


def predict_iris(sepal_length, sepal_width, petal_length, petal_width):
    input_data = np.array([[sepal_length, sepal_width, petal_length, petal_width]])
    prediction = model.predict(input_data)
    species = target_names[prediction[0]]

    # Return formatted prediction result
    return f"🎉 Predicted Species: {species}"


app = gr.Interface(
    fn=predict_iris,
    inputs=[
        gr.Slider(minimum=4.0, maximum=8.0, step=0.1, value=5.4, label="Sepal Length (cm)"),
        gr.Slider(minimum=2.0, maximum=4.5, step=0.1, value=3.4, label="Sepal Width (cm)"),
        gr.Slider(minimum=1.0, maximum=7.0, step=0.1, value=1.3, label="Petal Length (cm)"),
        gr.Slider(minimum=0.1, maximum=2.5, step=0.1, value=0.2, label="Petal Width (cm)")
    ],
    outputs=gr.Textbox(label="Real-time Prediction Output"),
    title="🌸 Iris Flower Interactive ML Web App",
    description="Adjust the sliders below to feed data into your trained machine learning model and view real-time predictions instantly.",
    theme="default"
)


app.launch(inline=True)
```
