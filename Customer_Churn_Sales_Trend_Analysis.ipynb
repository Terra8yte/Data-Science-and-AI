```python

# ============================================================
# CUSTOMER CHURN / SALES TREND ANALYSIS
# InternCircle Project
# Dataset: UCI Online Retail
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

print("=" * 65)
print("CUSTOMER CHURN / SALES TREND ANALYSIS")
print("=" * 65)

# 1. LOAD DATASET DIRECTLY FROM URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx"

try:
    df = pd.read_excel(url)
    print("\nDataset loaded successfully from URL!")
except Exception as e:
    print("\nDirect URL could not be loaded.")
    print("Trying the UCI dataset API...")
    %pip -q install ucimlrepo
    from ucimlrepo import fetch_ucirepo
    online_retail = fetch_ucirepo(id=352)
    df = online_retail.data.features.copy()
    print("Dataset loaded through UCI API!")

df.columns = df.columns.str.strip()

print("\nDataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

# 2. DATA UNDERSTANDING
print("\nDATASET INFORMATION")
df.info()

print("\nSTATISTICAL SUMMARY")
display(df.describe())

print("\nMISSING VALUES BEFORE CLEANING")
print(df.isnull().sum())

# 3. DATA CLEANING
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="coerce")

df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")
df["CustomerID"] = pd.to_numeric(df["CustomerID"], errors="coerce")

# Remove rows without key transaction fields or customer ID
df = df.dropna(subset=[
    "InvoiceDate", "CustomerID", "Quantity", "UnitPrice"
]).copy()

# Remove cancelled invoices and non-positive sales lines
df["InvoiceNo"] = df["InvoiceNo"].astype(str)
df = df[~df["InvoiceNo"].str.startswith("C")]
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]

df["CustomerID"] = df["CustomerID"].astype(int).astype(str)

# Calculate revenue
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print("\nDATASET SHAPE AFTER CLEANING:", df.shape)
print("\nMISSING VALUES AFTER CLEANING")
print(df.isnull().sum())

# 4. FEATURE ENGINEERING
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Month_Name"] = df["InvoiceDate"].dt.strftime("%b")
df["Quarter"] = df["InvoiceDate"].dt.to_period("Q").astype(str)
df["Day"] = df["InvoiceDate"].dt.day
df["Day_Name"] = df["InvoiceDate"].dt.day_name()
df["Hour"] = df["InvoiceDate"].dt.hour

print("\nFEATURE ENGINEERING COMPLETED")
display(df[[
    "InvoiceDate", "Year", "Month", "Month_Name",
    "Quarter", "Day", "Day_Name", "Hour", "Revenue"
]].head())

# 5. CUSTOMER-LEVEL ANALYSIS
analysis_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

customer_data = df.groupby("CustomerID").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Orders=("InvoiceNo", "nunique"),
    Total_Quantity=("Quantity", "sum"),
    Last_Order_Date=("InvoiceDate", "max"),
    Average_Order_Value=("Revenue", "mean"),
    Total_Products=("StockCode", "nunique")
).reset_index()

customer_data["Recency_Days"] = (
    analysis_date - customer_data["Last_Order_Date"]
).dt.days

print("\nCUSTOMER-LEVEL ANALYSIS")
display(customer_data.head())

# 6. CUSTOMER SEGMENTATION
# Activity-based proxy; this dataset has no confirmed churn label.

def customer_segment(row):
    if row["Recency_Days"] <= 90 and row["Total_Orders"] >= 3:
        return "Loyal / Active"
    elif row["Recency_Days"] <= 90:
        return "Recent Customer"
    elif row["Recency_Days"] > 180:
        return "At-Risk / Inactive"
    else:
        return "Needs Attention"

customer_data["Customer_Segment"] = customer_data.apply(
    customer_segment, axis=1
)

segment_summary = customer_data.groupby("Customer_Segment").agg(
    Customers=("CustomerID", "count"),
    Total_Revenue=("Total_Revenue", "sum"),
    Average_Revenue=("Total_Revenue", "mean"),
    Average_Orders=("Total_Orders", "mean"),
    Average_Recency=("Recency_Days", "mean")
).round(2)

print("\nCUSTOMER SEGMENTATION")
display(segment_summary)

# 7. MONTHLY SALES TREND
monthly_sales = (
    df.set_index("InvoiceDate")
      .resample("MS")["Revenue"]
      .sum()
      .reset_index()
)

monthly_sales["Year"] = monthly_sales["InvoiceDate"].dt.year
monthly_sales["Month"] = monthly_sales["InvoiceDate"].dt.month
monthly_sales["Month_Name"] = monthly_sales["InvoiceDate"].dt.strftime("%b")

monthly_sales["Previous_Month_Sales"] = monthly_sales["Revenue"].shift(1)
monthly_sales["MoM_Change_Percent"] = (
    monthly_sales["Revenue"].pct_change() * 100
).round(2)

print("\nMONTHLY SALES TREND")
display(monthly_sales)

# 8. PIVOT TABLES
country_pivot = pd.pivot_table(
    df,
    values="Revenue",
    index="Country",
    columns="Year",
    aggfunc="sum",
    fill_value=0
).round(2)

quarter_pivot = pd.pivot_table(
    df,
    values="Revenue",
    index="Year",
    columns="Quarter",
    aggfunc="sum",
    fill_value=0
).round(2)

customer_country_pivot = pd.pivot_table(
    df,
    values="Revenue",
    index="Country",
    aggfunc="sum",
    fill_value=0
).sort_values("Revenue", ascending=False).head(10).round(2)

print("\nREVENUE BY COUNTRY AND YEAR")
display(country_pivot)

print("\nREVENUE BY YEAR AND QUARTER")
display(quarter_pivot)

print("\nTOP 10 COUNTRIES BY REVENUE")
display(customer_country_pivot)

# 9. TOP PRODUCTS
top_products = (
    df.groupby("Description")
      .agg(
          Total_Quantity=("Quantity", "sum"),
          Total_Revenue=("Revenue", "sum")
      )
      .sort_values("Total_Revenue", ascending=False)
      .head(10)
      .reset_index()
)

print("\nTOP 10 PRODUCTS BY REVENUE")
display(top_products)

# 10. VISUALIZATION: MONTHLY REVENUE
plt.figure(figsize=(12, 5))
sns.lineplot(
    data=monthly_sales,
    x="InvoiceDate",
    y="Revenue",
    marker="o"
)
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (£)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 11. VISUALIZATION: TOP 10 PRODUCTS
plt.figure(figsize=(11, 6))
sns.barplot(
    data=top_products,
    x="Total_Revenue",
    y="Description"
)
plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue (£)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()

# 12. VISUALIZATION: TOP COUNTRIES
top_countries = (
    df.groupby("Country")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=top_countries,
    x="Revenue",
    y="Country"
)
plt.title("Top 10 Countries by Revenue")
plt.xlabel("Revenue (£)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()

# 13. VISUALIZATION: CUSTOMER SEGMENTS
segment_counts = (
    customer_data["Customer_Segment"]
    .value_counts()
    .rename_axis("Customer_Segment")
    .reset_index(name="Customers")
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=segment_counts,
    x="Customer_Segment",
    y="Customers"
)
plt.title("Customer Count by Activity Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# 14. VISUALIZATION: REVENUE BY CUSTOMER SEGMENT
segment_revenue = (
    customer_data.groupby("Customer_Segment")["Total_Revenue"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=segment_revenue,
    x="Customer_Segment",
    y="Total_Revenue"
)
plt.title("Revenue by Customer Activity Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Revenue (£)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# 15. VISUALIZATION: ORDER QUANTITY DISTRIBUTION
plt.figure(figsize=(9, 5))
sns.histplot(
    data=df[df["Quantity"] <= df["Quantity"].quantile(0.99)],
    x="Quantity",
    bins=40,
    kde=True
)
plt.title("Distribution of Order Quantities (up to 99th percentile)")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 16. VISUALIZATION: CORRELATION HEATMAP
numeric_columns = ["Quantity", "UnitPrice", "Revenue"]
correlation = df[numeric_columns].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

# 17. BUSINESS PERFORMANCE SUMMARY
total_revenue = df["Revenue"].sum()
total_orders = df["InvoiceNo"].nunique()
total_customers = df["CustomerID"].nunique()
total_quantity = df["Quantity"].sum()
average_order_value = total_revenue / total_orders if total_orders else 0

print("\n" + "=" * 65)
print("BUSINESS PERFORMANCE SUMMARY")
print("=" * 65)
print(f"Total Revenue:       £{total_revenue:,.2f}")
print(f"Total Orders:        {total_orders:,}")
print(f"Total Customers:     {total_customers:,}")
print(f"Units Sold:          {total_quantity:,.0f}")
print(f"Average Order Value: £{average_order_value:,.2f}")

# 18. KEY FINDINGS
top_product = top_products.iloc[0]["Description"]
top_product_revenue = top_products.iloc[0]["Total_Revenue"]

top_country = top_countries.iloc[0]["Country"]
top_country_revenue = top_countries.iloc[0]["Revenue"]

top_customer_segment = segment_summary["Total_Revenue"].idxmax()
at_risk_customers = customer_data[
    customer_data["Customer_Segment"] == "At-Risk / Inactive"
].shape[0]

best_month_index = monthly_sales["Revenue"].idxmax()
best_month = monthly_sales.loc[best_month_index, "InvoiceDate"].strftime("%B %Y")
best_month_revenue = monthly_sales.loc[best_month_index, "Revenue"]

print("\n" + "=" * 65)
print("KEY FINDINGS")
print("=" * 65)
print(f"Top product by revenue: {top_product}")
print(f"Product revenue: £{top_product_revenue:,.2f}")
print(f"Top country by revenue: {top_country}")
print(f"Country revenue: £{top_country_revenue:,.2f}")
print(f"Highest-revenue customer activity segment: {top_customer_segment}")
print(f"At-risk / inactive customer proxy count: {at_risk_customers}")
print(f"Highest-revenue month: {best_month}")
print(f"Monthly revenue: £{best_month_revenue:,.2f}")

# 19. ACTIONABLE BUSINESS RECOMMENDATIONS
print("\n" + "=" * 65)
print("ACTIONABLE BUSINESS RECOMMENDATIONS")
print("=" * 65)

print(f"""
1. PRODUCT STRATEGY:
   Review the product "{top_product}", which generated the highest
   revenue in this analysis. Monitor stock levels and evaluate
   product profitability before expanding inventory.

2. CUSTOMER RETENTION:
   {at_risk_customers} customers fall into the at-risk / inactive
   proxy group based on their last recorded purchase being over
   180 days before the analysis date. Consider a re-engagement
   campaign and measure its return rate.

3. CUSTOMER LOYALTY:
   Study the purchasing behaviour of the {top_customer_segment}
   segment. Consider loyalty offers or personalized product
   recommendations and evaluate their impact.

4. REGIONAL STRATEGY:
   {top_country} generated the highest revenue. Examine customer
   demand and product preferences in this market before adapting
   those practices to other countries.

5. REVENUE PLANNING:
   The highest-revenue month was {best_month}. Compare monthly
   patterns to plan inventory and promotional activity, while
   accounting for the dataset's limited date range.

6. DATA-DRIVEN DECISIONS:
   Review revenue, order frequency and customer recency together.
   Revenue alone does not fully describe customer loyalty or
   long-term profitability.
""")

# 20. PROJECT CONCLUSION
print("=" * 65)
print("PROJECT CONCLUSION")
print("=" * 65)

print("""
The Online Retail dataset was analyzed using Pandas,
Matplotlib and Seaborn. The project included data cleaning,
date-based feature engineering, customer-level metrics,
activity-based customer segmentation, pivot tables,
sales trend analysis and visualizations.

The analysis highlights revenue patterns, product performance
and customer activity. The inactive-customer segment is only
a churn-risk proxy because the dataset does not provide a
confirmed churn label. Business recommendations should be
tested with subsequent transaction data.

PROJECT COMPLETED SUCCESSFULLY!
""")
```
